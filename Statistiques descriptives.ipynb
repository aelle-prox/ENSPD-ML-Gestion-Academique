```python
import pandas as pd

# Charger le dataframe (exemple)
df = pd.read_csv("ensp_dataset_5000_clean.csv")  # modifie le nom si besoin

# Sélectionner uniquement les colonnes numériques
df_num = df.select_dtypes(include=["int64", "float64"])

# Calcul des statistiques descriptives
stats = pd.DataFrame({
    "Min": df_num.min(),
    "Max": df_num.max(),
    "Moyenne": df_num.mean(),
    "Médiane": df_num.median(),
    "Écart-type": df_num.std()
})

# Affichage des résultats
print(stats)
```
