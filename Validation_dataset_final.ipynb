```python
import pandas as pd

# Chargement du dataset
df = pd.read_csv("dataset_final_pret_ML.csv")

print("\n===== APERÇU DU DATASET =====")
print(df.head())
print("Dimensions :", df.shape)

print("\n===== COLONNES DISPONIBLES =====")
print(df.columns.tolist())

# Valeurs manquantes
print("\n===== VALEURS MANQUANTES =====")
missing = df.isnull().sum()
print(missing)

# Doublons
print("\n===== DOUBLONS =====")
duplicates = df.duplicated().sum()
print("Nombre de doublons :", duplicates)

# Incohérences
print("\n===== INCOHÉRENCES LOGIQUES =====")

# Âge
if "age" in df.columns:
    age_errors = df[(df["age"] < 15) | (df["age"] > 40)]
    print("Âges incohérents :", len(age_errors))
else:
    print("Colonne 'age' absente")

# Note (plusieurs noms possibles)
note_columns = ["note_moyenne", "note", "score", "moyenne"]

note_found = False
for col in note_columns:
    if col in df.columns:
        note_errors = df[(df[col] < 0) | (df[col] > 100)]
        print(f"Notes incohérentes ({col}) :", len(note_errors))
        note_found = True
        break

if not note_found:
    print("Aucune colonne de notes trouvée")

# Présence
presence_columns = ["taux_presence", "presence", "assiduite"]

presence_found = False
for col in presence_columns:
    if col in df.columns:
        presence_errors = df[(df[col] < 0) | (df[col] > 100)]
        print(f"Taux de présence incohérents ({col}) :", len(presence_errors))
        presence_found = True
        break

if not presence_found:
    print("Aucune colonne de présence trouvée")

# Résumé final
print("\n===== RÉSUMÉ FINAL =====")
print("Valeurs manquantes totales :", missing.sum())
print("Doublons :", duplicates)

print("\n✅ ÉTAPE 1 DU JOUR 5 TERMINÉE")
print("👉 Dataset validé pour passer au feature engineering")
```
