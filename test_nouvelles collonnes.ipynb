```python
import pandas as pd

# Chargement du dataset nettoyé
df = pd.read_csv("ensp_dataset_5000_clean.csv")
print("Dataset chargé :", df.shape)

# Afficher les colonnes existantes pour référence
print("\nColonnes disponibles :")
print(df.columns.tolist())

# Liste pour suivre les nouvelles colonnes créées
nouvelles_colonnes = []

# 1️⃣ Création du ratio présence / note
presence_col = None
note_col = None

# Détection automatique des colonnes
for col in df.columns:
    if col.lower() in ["taux_presence", "presence", "assiduite"]:
        presence_col = col
    if col.lower() in ["note_moyenne", "note", "score", "moyenne"]:
        note_col = col

if presence_col and note_col:
    df["ratio_presence_note"] = df[presence_col] / df[note_col]
    nouvelles_colonnes.append("ratio_presence_note")
    print("✅ Ratio présence/note créé")
else:
    print("⚠️ Colonnes nécessaires pour ratio présence/note absentes")

# 2️⃣ Moyenne par filière
filiere_col = None
for col in df.columns:
    if col.lower() == "filiere":
        filiere_col = col
        break

if filiere_col and note_col:
    df["moyenne_filiere"] = df.groupby(filiere_col)[note_col].transform("mean")
    nouvelles_colonnes.append("moyenne_filiere")
    print("✅ Moyenne par filière créée")
else:
    print("⚠️ Colonnes nécessaires pour moyenne par filière absentes")

# 3️⃣ Engagement académique
if presence_col and note_col:
    df["engagement_academique"] = (df[presence_col] + df[note_col]) / 2
    nouvelles_colonnes.append("engagement_academique")
    print("✅ Engagement académique créé")
else:
    print("⚠️ Colonnes nécessaires pour engagement académique absentes")

# 4️⃣ Aperçu des nouvelles colonnes créées
if nouvelles_colonnes:
    print("\n=== Aperçu des nouvelles variables ===")
    print(df[nouvelles_colonnes].head())
else:
    print("\n⚠️ Aucune nouvelle variable n’a été créée, vérifier les noms des colonnes !")

# 5️⃣ Sauvegarde du dataset enrichi
df.to_csv("ensp_dataset_5000_feature_engineered.csv", index=False)
print("\n✅ Dataset sauvegardé : ensp_dataset_5000_feature_engineered.csv")
```
