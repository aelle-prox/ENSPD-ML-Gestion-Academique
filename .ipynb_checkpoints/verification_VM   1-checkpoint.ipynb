```python
# =========================================
# JOUR 4 – ANALYSE DES VALEURS MANQUANTES
# (SANS IMPUTATION)
# =========================================

import pandas as pd
import matplotlib.pyplot as plt

# 1️⃣ Charger le fichier CSV (même dossier)
df = pd.read_csv("ensp_dataset_5000.csv")

print("✅ Fichier chargé avec succès")
print("\nAperçu des données :")
print(df.head())

# 2️⃣ Informations générales sur le dataset
print("\n📊 Informations générales :")
print(df.info())

# 3️⃣ Calcul des valeurs manquantes
missing_count = df.isnull().sum()
missing_percent = (df.isnull().mean() * 100).round(2)

missing_df = pd.DataFrame({
    "Nombre_valeurs_manquantes": missing_count,
    "Pourcentage_%": missing_percent
})

# Garder uniquement les colonnes avec des valeurs manquantes
missing_df = missing_df[missing_df["Nombre_valeurs_manquantes"] > 0]
missing_df = missing_df.sort_values(by="Pourcentage_%", ascending=False)

print("\n📌 Tableau des valeurs manquantes :")
print(missing_df)

# 4️⃣ Visualisation simple (bar chart)
plt.figure()
missing_df["Pourcentage_%"].plot(kind='bar')
plt.title("Pourcentage de valeurs manquantes par variable")
plt.ylabel("Pourcentage (%)")
plt.xlabel("Variables")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()

# 5️⃣ Nombre de lignes incomplètes
lignes_incompletes = df.isnull().any(axis=1).sum()
print(f"\n📉 Nombre de lignes contenant au moins une valeur manquante : {lignes_incompletes}")

# 6️⃣ Sauvegarde du rapport des valeurs manquantes
missing_df.to_csv("rapport_valeurs_manquantes.csv")

print("\n📁 Rapport sauvegardé : rapport_valeurs_manquantes.csv")
```
