```python
# ================================
# JOUR 4 – NETTOYAGE DES DONNÉES
# ================================

import pandas as pd

# 1️⃣ Charger le fichier CSV (même dossier)
df = pd.read_csv("ensp_dataset_5000.csv")

print("✅ Fichier chargé avec succès !")
print("\nAperçu des données :")
print(df.head())

# 2️⃣ Vérifier les valeurs manquantes
missing_count = df.isnull().sum()
missing_percent = (df.isnull().mean() * 100).round(2)

missing_df = pd.DataFrame({
    "Nombre_valeurs_manquantes": missing_count,
    "Pourcentage_%": missing_percent
}).sort_values(by="Pourcentage_%", ascending=False)

print("\n📌 Valeurs manquantes par colonne :")
print(missing_df)

# ================================
# OPTION 1 – IMPUTATION SIMPLE
# ================================

# 3️⃣ Colonnes numériques → médiane
colonnes_numeriques = df.select_dtypes(include=['int64', 'float64']).columns

for col in colonnes_numeriques:
    df[col].fillna(df[col].median(), inplace=True)

# 4️⃣ Colonnes catégorielles → mode
colonnes_categorielles = df.select_dtypes(include=['object']).columns

for col in colonnes_categorielles:
    df[col].fillna(df[col].mode()[0], inplace=True)

# 5️⃣ Vérification après imputation
print("\n✅ Valeurs manquantes après imputation :")
print(df.isnull().sum())

# 6️⃣ Sauvegarder le fichier nettoyé
df.to_csv("ensp_dataset_5000_clean.csv", index=False)

print("\n📁 Fichier nettoyé sauvegardé : ensp_dataset_5000_clean.csv")
```
