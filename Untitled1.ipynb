```python
import streamlit as st
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import pickle

# Charger les modèles sauvegardés
with open("log_model.pkl", "rb") as f:
    log_model = pickle.load(f)

with open("tree_model.pkl", "rb") as f:
    tree_model = pickle.load(f)

st.title("💻 Prédiction de réussite académique")

# -------------------
# Sidebar : Entrées
# -------------------
st.sidebar.header("Informations de l'étudiant")
age = st.sidebar.number_input("Âge", min_value=16, max_value=100, value=20)
year_of_study = st.sidebar.number_input("Année d'étude", 1, 10, 2)
attendance_rate = st.sidebar.slider("Taux de présence", 0.0, 1.0, 0.9)
assignments_avg = st.sidebar.number_input("Moyenne devoirs", 0, 20, 15)
midterm_score = st.sidebar.number_input("Score partiel", 0, 20, 14)
final_score = st.sidebar.number_input("Score final", 0, 20, 16)
project_participation = st.sidebar.slider("Participation projets", 0.0, 1.0, 0.8)

# Ici, ajouter d'autres variables si nécessaire (simplification pour l'exemple)

model_choice = st.selectbox("Choisir le modèle", ["Régression Logistique", "Arbre de Décision"])

# -------------------
# Bouton de prédiction
# -------------------
if st.button("Prédire la réussite"):
    # Préparer les données
    X_new = np.array([age, year_of_study, attendance_rate,
                      assignments_avg, midterm_score, final_score,
                      project_participation]).reshape(1, -1)
    
    if model_choice == "Régression Logistique":
        pred = log_model.predict(X_new)[0]
        proba = log_model.predict_proba(X_new)[0]
    else:
        pred = tree_model.predict(X_new)[0]
        proba = tree_model.predict_proba(X_new)[0]
    
    st.success(f"Prédiction : {'Réussite' if pred==1 else 'Échec'}")
    
    # Graphique de probabilité
    fig, ax = plt.subplots()
    ax.bar(["Échec", "Réussite"], proba, color=["red","green"])
    ax.set_ylim(0,1)
    ax.set_ylabel("Probabilité")
    ax.set_title("Probabilité de réussite")
    st.pyplot(fig)
```
